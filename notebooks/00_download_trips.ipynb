# 00 · Download the Bluebikes trips

**Who Gets a Bluebike?** · CP 101 final project

The monthly trip files are too large for GitHub (37 to 134 MB each), so this notebook downloads them from Bluebikes into `data/raw/bluebikes/`.
Run it once. A file that is already there is skipped, so running it again is safe.

The census data and the boundary files are small, so they are already in the repository and need no download.

```python
from pathlib import Path
import urllib.request
import zipfile

STUDY_YEAR = 2024
RAW_TRIPS = Path("../data/raw/bluebikes")
RAW_TRIPS.mkdir(parents=True, exist_ok=True)   # create the folder if it does not exist yet
```

Bluebikes publishes one zip file per month, named like `202401-bluebikes-tripdata.zip`.
Each zip holds the CSV and a hidden macOS file called `__MACOSX/._...`, so we extract only the CSV, by its exact name.

```python
for month in range(1, 13):
    name = str(STUDY_YEAR) + str(month).zfill(2) + "-bluebikes-tripdata"   # e.g. "202401-bluebikes-tripdata"
    csv_path = RAW_TRIPS / (name + ".csv")

    if csv_path.exists():
        print("already downloaded:", csv_path.name)
    else:
        url = "https://s3.amazonaws.com/hubway-data/" + name + ".zip"
        with urllib.request.urlopen(url) as response:
            zip_bytes = response.read()

        zip_path = RAW_TRIPS / (name + ".zip")
        zip_path.write_bytes(zip_bytes)          # save the zip file

        with zipfile.ZipFile(zip_path) as archive:
            archive.extract(name + ".csv", RAW_TRIPS)   # take out only the CSV

        zip_path.unlink()                        # delete the zip; we only need the CSV
        print("downloaded:", csv_path.name)

print(len(list(RAW_TRIPS.glob("*.csv"))), "trip files in", RAW_TRIPS)
```
